Correr desde la raíz del TP con `venv/bin/jupyter lab ex3_nglview.ipynb`. Las celdas usan la misma carga de estructuras que el resto del TP (`structures.load`), así que los archivos salen de `data/`.

```python
import nglview as nv
from structures import load

mb = load('1A6M')     # oximioglobina
ake = load('1AKE')    # adenilato quinasa cerrada, con Ap5A
```

## Representación por defecto

```python
view = nv.show_biopython(mb)
view
```

## Cartoon + hemo, O2 y His proximal/distal como bastones

En la selección de NGL, `HEM`/`OXY` son nombres de residuo y `93`/`64` números de residuo: His93 es la proximal (está unida al Fe, según el registro `LINK` del archivo), His64 la distal (su átomo NE2 está a 2,97 Å del O2, distancia típica de un puente de hidrógeno; lo medimos en Python).

```python
v = nv.show_biopython(mb)
v.clear_representations()
v.add_representation('cartoon', selection='protein', color='lightgrey')
v.add_representation('licorice', selection='HEM', color='yellow')
v.add_representation('licorice', selection='OXY', color='blue')
v.add_representation('licorice', selection='93 or 64', color='green')
v.add_representation('spacefill', selection='_FE', color='orange', radius=0.6)
v
```

## Pocos residuos aislados

En la lisozima (1LSE), sobre su superficie en transparencia: todos los aspárticos (`ASP`) en rojo y los glutamatos (`GLU`) en azul, para ver cuáles quedan en la superficie y cuáles adentro. Los aspárticos de la lisozima son exactamente los 7 de la tabla del ejercicio 6.

```python
lys = load('1LSE')
v = nv.show_biopython(lys)
v.clear_representations()
v.add_representation('surface', selection='protein', opacity=0.25, color='white')
v.add_representation('ball+stick', selection='ASP', color='red')
v.add_representation('ball+stick', selection='GLU', color='blue')
# v.add_representation('ball+stick', selection='ASP and (18 or 48 or 52 or 66 or 87 or 101 or 119)', color='red')
v
```

## Adenilato quinasa coloreada por dominio

CORE en gris, NMP (30–59) en azul, LID (122–159) en rojo (límites de dominio tomados de la literatura sobre la enzima, no del archivo), Ap5A como bastones: en la forma cerrada la tapa se pliega sobre el inhibidor.

```python
v = nv.show_biopython(ake)
v.clear_representations()
v.add_representation('cartoon', selection=':A', color='grey')
v.add_representation('cartoon', selection=':A and 30-59', color='blue')
v.add_representation('cartoon', selection=':A and 122-159', color='red')
v.add_representation('licorice', selection=':A and AP5', color='yellow')
v
```
