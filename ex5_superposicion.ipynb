## Mioglobina de cachalote y humana, superpuestas

Las dos mioglobinas difieren en 21 de las 149 posiciones comparadas, pero su forma 3D es prácticamente la misma. Primero las superponemos con el mismo código del ejercicio 5 (la de cachalote queda fija, la humana se mueve), y después las dibujamos juntas: cachalote en celeste, humana en salmón, y los aminoácidos que difieren como bastones (azul en cachalote, rojo en humana).

Ojo: 3RGK es un mutante (K45R), así que en la posición 45 es igual al cachalote solo por la mutación; con la mioglobina humana natural serían 22 diferencias. Además, a 3RGK le faltan los aminoácidos 150–153, que el experimento no logró resolver.

```python
from exercises.ex5.ex5 import equivalent_cas
from geometry import superimpose

whale = load('1A6M')   # cachalote: referencia, no se mueve
human = load('3RGK')   # humano (mutante K45R): se mueve sobre la de cachalote

aln, pairs = equivalent_cas(whale[0]['A'], human[0]['A'])
sup = superimpose([w for w, h in pairs], [h for w, h in pairs], human)

# Posiciones alineadas donde el aminoácido es distinto, con el número de cada una en su propio archivo.
different = [(w.get_parent(), h.get_parent()) for w, h in pairs
             if w.get_parent().get_resname() != h.get_parent().get_resname()]
print(f'{len(pairs)} posiciones comparadas, {len(different)} distintas; '
      f'RMSD después de superponer: {sup.rms:.2f} Å')
for w, h in different:
    print(f'  cachalote {w.get_resname()}{w.id[1]:<4} humano {h.get_resname()}{h.id[1]}')
```

```python
def numbers(residues):
    return ' or '.join(str(r.id[1]) for r in residues)

v = nv.NGLWidget()
v.add_component(nv.BiopythonStructure(whale))   # componente 0
v.add_component(nv.BiopythonStructure(human))   # componente 1
for c in (0, 1):
    v.clear_representations(component=c)

v.add_representation('cartoon', selection='protein', color='skyblue', component=0)
v.add_representation('cartoon', selection='protein', color='salmon', component=1)
v.add_representation('licorice', selection=f'protein and ({numbers(w for w, h in different)})',
                     color='blue', component=0)
v.add_representation('licorice', selection=f'protein and ({numbers(h for w, h in different)})',
                     color='red', component=1)
v.add_representation('licorice', selection='HEM', component=0)
v
```
